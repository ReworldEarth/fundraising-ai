```python
import pandas as pd
import jsonlines
import json
```

```python
file_path = '../data/supercleaned_for_embedding.jsonl'

with jsonlines.open(file_path) as reader:
    for obj in reader:
        print(obj)  # Process each JSON object
```

```python
file_path = '../data/supercleaned_for_embedding.jsonl'
data = [] # Create an empty list to store the JSON objects
with open(file_path, 'r') as f:
    for line in f:
        try:
            json_object = json.loads(line) # Parse each line as a JSON object
            data.append(json_object) # Append the JSON object to the list
        except json.JSONDecodeError as e:
            print(f"Error decoding JSON on line: {e}")
            continue
```
